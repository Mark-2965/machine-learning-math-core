```python
# Let's create a new column called `price_category` based on the `price`
# column. We'll categorize wines into 'Budget' (price < 20),
# 'Mid-Range' (20 <= price < 50), and 'Premium' (price => 50).
```

```python
df['country'].head(10)
```

```python
country_map = {
    "US" : "United States",
    "France" : "FR",
    "Italy" : "IT",
    "Spain": "SP"
}
```

```python
df["country_new"] = df["country"].map(country_map)
```

```python
df[["country","country_new"]].head(20)
```

```python
df["price_ngn"] = df["price"] * 1600
```

```python
df[["country","country_new","price","price_ngn"]]
```

```python
df[["price","price_ngn"]].head()
```

```python
# Suppose you want Expensive if price > 50
# Cheap otherwise
```

```python
df["category"] = df["price"]. map(lambda x : "Expensive" if x > 50 else "Cheap")
```

```python
df[["country_new","price","category"]].head(10)
```

```python
df["categorize_price"] = df["price"].map(lambda x: "Unknown" if pd.isna(x) else ("Budget"
                                                                                 if x < 20 else ("Mid-Range" if 20 <= x < 50 else "Premium")))
```

```python
df["categorize_price"]
```

```python
def categorize_price(price):
    if pd.isna(price):
        return "Unknown"
    elif price < 20:
        return "Budget"
    elif 20 <= price < 50:
        return "Mid-Range"
    else:
        return "Premium"
df["price_category"] = df["price"].apply(categorize_price)
```

```python
import pandas as pd
df = pd.read_csv("winemag-data-130k-v2.csv")
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define the function to categorize price
def categorize_price(price):
    if pd.isna(price):
        return "Unknown"
    elif price < 20:
        return "Budget"
    elif 20 <= price < 50:
        return "Mid-Range"
    else:
        return "Premium"

# Apply the function to create the 'price_category' column
df["price_category"] = df["price"].apply(categorize_price)

# Create a scatter plot of points vs price
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='points', y='price', alpha=0.9, hue='price_category', palette='viridis')
plt.title('Wine Points vs Price')
plt.xlabel('Points')
plt.ylabel('Price')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='country', y='price', alpha=0.9, palette='viridis')
plt.title('Country VS Price')
plt.xlabel('Country')
plt.ylabel('Price')
plt.show()
```

```python
# Create a bar plot of the "price_category" distribution

plt.figure(figsize=(3, 6))
sns.countplot(data=df, x='price_category', order=['Budget', 'Mid-Range', 'Premium', 'Unknown'], palette='viridis', hue='price_category', legend=False)
plt.title('Distribution of Wine Price Categories')
plt.xlabel('Price Category')
plt.ylabel('Number of Wines')
plt.show()
```

```python
# Create a bar plot of the 'country' distribution

plt.figure(figsize=(8, 6))
sns.countplot(data=df, x='country', palette='viridis', hue='country', legend=False)
plt.title('Distribution of Wines by Country')
plt.xlabel('Country')
plt.ylabel('Number of Wines')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Bar chart for 'country' (showing top 10 countries for readability)
plt.figure(figsize=(12, 6))
top_countries = df['country'].value_counts().nlargest(10).index
sns.countplot(data=df[df['country'].isin(top_countries)], x='country', order=top_countries, palette='viridis')
plt.title('Top 10 Countries by Number of Wines')
plt.xlabel('Country')
plt.ylabel('Number of Wines')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Histogram for 'price'
plt.figure(figsize=(10, 6))
sns.histplot(df['price'].dropna(), bins=30, kde=True, color='skyblue')
plt.title('Distribution of Wine Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Bar chart for 'winery' (showing top 10 wineries for readability)
plt.figure(figsize=(12, 6))
top_wineries = df['winery'].value_counts().nlargest(10).index
sns.countplot(data=df[df['winery'].isin(top_wineries)], x='winery', order=top_wineries, palette='magma')
plt.title('Top 10 Wineries by Number of Wines')
plt.xlabel('Winery')
plt.ylabel('Number of Wines')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
