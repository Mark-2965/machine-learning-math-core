```python
import numpy as np
import pandas as pd
import xgboost as xgb

print("🤖 INITIALIZING BISMARK DATA LABS ENTERPRISE XGBOOST MODEL CORE ON THE CLOUD...\n")

# 📊 1. PREPARE THE Tabular DATASET (Lekki / Ajah Case Study Metrics)
# Features: [Square Footage, Number of Bedrooms, Distance to VGC in Kilometers]
sizes = [1000, 1500, 2000, 3000, 1200, 1700, 2500, 3500]
rooms = [2, 3, 3, 5, 2, 3, 4, 5]
vgc_distance = [5.0, 3.5, 2.0, 1.0, 4.5, 3.0, 1.5, 0.5]
prices = [30.0, 45.0, 55.0, 95.0, 35.0, 50.0, 75.0, 110.0]

raw_data = {
    'Square_Footage': sizes,
    'Bedrooms': rooms,
    'Distance_to_VGC': vgc_distance,
    'Price_Millions': prices
}

df = pd.DataFrame(raw_data)
print("📊 COMPLETE TREE-BOOSTING INDUSTRIAL TESTING DATASET LOADED:")
print(df)
print("-" * 75)

# Separate input feature vectors (X) from target values (y)
X = df[['Square_Footage', 'Bedrooms', 'Distance_to_VGC']]
y = df['Price_Millions']

# 🧠 2. INITIALIZE THE EXTREME GRADIENT BOOSTING REGRESSOR CORE
# n_estimators: The maximum number of sequential correction trees to build
# learning_rate: The scaling dampener factor (shrinkage) applied to each subsequent tree's fix
# max_depth: How deep each flowchart tree can branch out to separate non-linear conditions
model = xgb.XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

# 🏋️ 3. TRAIN THE ENSEMBLE TREE CORRECTION PIPELINE (Boosting Loops)
model.fit(X, y)
print("✅ XGBOOST TREE TRAINING MATRIX LOCKED IN.")
print("-" * 75)

# 🔮 4. INFERENCE: RUN A PREDICTION VALUATION ON A NEW TARGET PROPERTY
# Let's check a brand new house layout: 2,200 sq ft, 4 Bedrooms, sitting 1.8km from VGC!
new_house = pd.DataFrame(
    [[2200, 4, 1.8]],
    columns=['Square_Footage', 'Bedrooms', 'Distance_to_VGC']
)
predicted_price = model.predict(new_house)

print(f"🔮 [XGBOOST INDUSTRIAL INFERENCE RESULT]:")
print(f"For a brand new 2,200 sq ft, 4-bed property sitting 1.8km from VGC:")
print(f"➔ XGBoost AI Calculated Valuation: ₦{predicted_price[0]:.2f} Million Naira!")
print("-" * 75)
```
