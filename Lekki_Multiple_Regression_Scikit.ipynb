```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# 📊 1. STRUCTURE THE DATASET USING A PROFESSIONAL PANDAS DATAFRAME
sizes = [1000, 1500, 2000, 3000]
rooms = [2, 3, 3, 5]
prices = [30.0, 45.0, 55.0, 95.0]

raw_data = {
    'Square_Footage': sizes,
    'Bedrooms': rooms,
    'Price_Millions': prices
}

df = pd.DataFrame(raw_data)
print("📊 TRAINING DATA MATRIX LOADED SUCCESSFULLY:")
print(df)
print("-" * 60)

# ✂️ 2. SEPARATE THE FEATURES (INPUTS) FROM THE TARGET (OUTPUT)
X_features = df[['Square_Footage', 'Bedrooms']]
y_target = df['Price_Millions']

# 🧠 3. INITIALIZE AND TRAIN THE ENTERPRISE AI MODEL
# Scikit-learn automatically manages intercepts and matrix transposes for us!
model = LinearRegression()
model.fit(X_features, y_target)

print("✅ MODEL TRAINING COMPLETE via SCIKIT-LEARN CORE ENGINE.")
print(f"• Base Intercept Price: ₦{model.intercept_ * 1000000:,.2f}")
print(f"• Size Coefficient:    ₦{model.coef_[0] * 1000000:,.2f} per sq ft")
print(f"• Room Coefficient:    ₦{model.coef_[1] * 1000000:,.2f} per bedroom")
print("-" * 60)

# 🔮 4. INFERENCE: PREDICT THE VALUE OF A 2500 SQ FT, 4 BEDROOM LEKKI HOUSE
new_house = pd.DataFrame([[2500, 4]], columns=['Square_Footage', 'Bedrooms'])
predicted_price = model.predict(new_house)[0]

print(f"🔮 [SCIKIT-LEARN CLOUD PREDICTION]:")
print(f"➔ AI Estimated Value: ₦{predicted_price:.2f} Million Naira!")
print("-" * 60)
```
